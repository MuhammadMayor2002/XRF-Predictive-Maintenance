# XRF Predictive Maintenance: Anomaly Detection

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define Columns
columns = [
    'Type', 'Day', 'Month', 'Year', 'Hour', 'Minute', 'Sample_Type', 'Mode', 
    'Empty1', 'Empty2', 'Empty3', 'Empty4', 'Empty5', 'Empty6', 'Sample_ID', 'Track_ID', 
    'Empty7', 'Empty8', 'Empty9', 'Empty10', 'Empty11', 'Empty12', 'Num_Elements', 
    'El1_Name', 'El1_Flag', 'El1_Value',  
    'El2_Name', 'El2_Flag', 'El2_Value',  
    'El3_Name', 'El3_Flag', 'El3_Value',  
    'El4_Name', 'El4_Flag', 'El4_Value'   
]

# 2. Load Data
df = pd.read_csv('ARL_ANALYSIS.CSV', header=None, names=columns, usecols=range(35), engine='python')

# 3. Fix the Date
date_string = (
    df['Year'].astype(str) + '-' + 
    df['Month'].astype(str).str.zfill(2) + '-' + 
    df['Day'].astype(str).str.zfill(2) + ' ' + 
    df['Hour'].astype(str).str.zfill(2) + ':' + 
    df['Minute'].astype(str).str.zfill(2)
)
df['Datetime'] = pd.to_datetime(date_string, format='%Y-%m-%d %H:%M', errors='coerce')

# 4. Filter for Clinker and Clean
clinker_df = df[df['Sample_Type'] == 'CLINKER_'].copy()
clinker_df = clinker_df.rename(columns={'El4_Value': 'CaO_Percentage'})
clinker_df['CaO_Percentage'] = pd.to_numeric(clinker_df['CaO_Percentage'], errors='coerce')
clinker_df = clinker_df[clinker_df['CaO_Percentage'] > 1.0]

# 5. Sort by Date
clinker_df = clinker_df.sort_values('Datetime')

# 6. Plot
plt.figure(figsize=(14, 6))
sns.scatterplot(data=clinker_df, x='Datetime', y='CaO_Percentage', color='blue', alpha=0.6, label='Measured CaO')

clinker_df.set_index('Datetime', inplace=True)
clinker_df['CaO_Rolling_Mean'] = clinker_df['CaO_Percentage'].rolling('7D').mean()

plt.plot(clinker_df.index, clinker_df['CaO_Rolling_Mean'], color='red', linewidth=3, label='7-Day Trend')

plt.title('XRF Instrument Drift: Clinker CaO Measurements Over Time')
plt.ylabel('CaO Percentage (%)')
plt.xlabel('Date')
plt.legend()
plt.grid(True)
plt.show()

# 7. Print verification
print("--- DATA VERIFICATION ---")
print(clinker_df[['Sample_Type', 'CaO_Percentage']].head())
```

## 1. Data Ingestion & Cleaning

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re

# 1. Read the raw file as plain text lines
with open('ARL_ANALYSIS.CSV', 'r') as file:
    lines = file.readlines()

data = []

# 2. Parse line by line
for line in lines:
    if 'CLINKER_' in line or 'Cement_F' in line:
        parts = line.split(',')
        
        try:
            # FIX DATE: Force the Year to be 2026 (since the file says '26')
            day = parts[1].strip().zfill(2)
            month = parts[2].strip().zfill(2)
            year = "2026" # Hardcoded based on your file
            hour = parts[4].strip().zfill(2)
            minute = parts[5].strip().zfill(2)
            
            date_str = f"{year}-{month}-{day} {hour}:{minute}"
            dt = pd.to_datetime(date_str, format='%Y-%m-%d %H:%M', errors='coerce')
            
            sample_type = "CLINKER_" if "CLINKER_" in line else "CEMENT_F"
            
            # FIX EXTRACTION: Find 'CaO' or 'Ca', then find the very next number in the list
            cao_val = None
            for i, part in enumerate(parts):
                clean_part = part.strip()
                if clean_part in ['CaO', 'Ca']:
                    # Look at the next 5 elements to find the first valid number
                    for j in range(i+1, i+6):
                        if j < len(parts):
                            val_str = parts[j].strip()
                            try:
                                cao_val = float(val_str)
                                break # Found it!
                            except ValueError:
                                continue # Not a number, keep looking
                    if cao_val is not None:
                        break
                        
            if cao_val is not None and not pd.isna(dt):
                data.append({'Datetime': dt, 'Sample_Type': sample_type, 'CaO': cao_val})
                
        except Exception as e:
            continue

# 3. Create DataFrame
df = pd.DataFrame(data)

if len(df) > 0:
    print(f"Successfully extracted {len(df)} valid samples!")
    
    # 4. Plot
    plt.figure(figsize=(14, 6))
    
    clinker_df = df[df['Sample_Type'] == 'CLINKER_'].sort_values('Datetime')
    
    # Plot Clinker
    if not clinker_df.empty:
        sns.scatterplot(data=clinker_df, x='Datetime', y='CaO', color='blue', alpha=0.6, label='Clinker CaO Readings')
        
        # Rolling trend
        clinker_df.set_index('Datetime', inplace=True)
        clinker_df['CaO_Trend'] = clinker_df['CaO'].rolling('2D').mean()
        plt.plot(clinker_df.index, clinker_df['CaO_Trend'], color='darkblue', linewidth=3, label='Clinker 2-Day Trend')

    plt.title('XRF Instrument Drift: Clinker CaO Measurements Over Time')
    plt.ylabel('Calcium Percentage (%)')
    plt.xlabel('Date')
    plt.legend()
    plt.grid(True)
    plt.show()
    
    # Print the data to verify
    print(clinker_df[['CaO']].head(10))
else:
    print("Could not extract data.")
```

## 2. Exploratory Data Analysis (CaO Drift)

```python
from sklearn.ensemble import IsolationForest

# 1. Prepare the data for the AI
# We will give the AI the CaO percentage to learn from
X = clinker_df[['CaO']].values

# 2. Initialize the Isolation Forest model
# We set contamination=0.05, meaning we expect about 5% of the data to be anomalies
model = IsolationForest(contamination=0.05, random_state=42)

# 3. Train the model and predict anomalies
clinker_df['Anomaly_Score'] = model.fit_predict(X)

# The model returns 1 for normal data, and -1 for anomalies
clinker_df['Is_Anomaly'] = clinker_df['Anomaly_Score'] == -1

# 4. Plot the results
plt.figure(figsize=(14, 6))

# Plot normal data
normal_data = clinker_df[clinker_df['Is_Anomaly'] == False]
plt.scatter(normal_data.index, normal_data['CaO'], color='green', label='Normal Operation', alpha=0.6)

# Plot anomalies (Red dots)
anomaly_data = clinker_df[clinker_df['Is_Anomaly'] == True]
plt.scatter(anomaly_data.index, anomaly_data['CaO'], color='red', s=100, label='AI Detected Anomaly', edgecolors='black')

plt.title('AI Anomaly Detection: Clinker CaO Measurement')
plt.ylabel('CaO Percentage (%)')
plt.xlabel('Date')
plt.legend()
plt.grid(True)
plt.show()

# 5. Print the dates where anomalies occurred
print("--- DATES FLAGGED AS ANOMALIES BY THE AI ---")
print(anomaly_data[['CaO']].to_string())
```

## 3. Machine Learning (Isolation Forest)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create fake Machine Status Data for the 38 real dates we have
np.random.seed(42)
n_samples = len(clinker_df)

# Simulate normal machine parameters
# SmplTime (Seconds it takes to pull vacuum) ~ normally 12 seconds
smpl_time = np.random.normal(12, 1, n_samples)
# Water Temp ~ normally 36 degrees
water_temp = np.random.normal(36, 0.5, n_samples)

# The Feb 12 anomaly (index 1 in our df) will have a MASSIVE vacuum failure
# We simulate that the pump took 140 seconds and water temp spiked to 45 degrees!
smpl_time[1] = 145.0 
water_temp[1] = 44.5

# 2. Merge the fake machine data into our real Clinker dataframe
clinker_df['SmplTime'] = smpl_time
clinker_df['WaterTemp'] = water_temp

# 3. Train the AI on BOTH chemistry AND machine health
features = ['CaO', 'SmplTime', 'WaterTemp']
X_multi = clinker_df[features].values

model_multi = IsolationForest(contamination=0.05, random_state=42)
clinker_df['Multi_Anomaly'] = model_multi.fit_predict(X_multi)
clinker_df['Is_Anomaly_Multi'] = clinker_df['Multi_Anomaly'] == -1

# 4. Plot the AI Dashboard
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

# Top Plot: Vacuum Pump Health
ax1.plot(clinker_df.index, clinker_df['SmplTime'], color='purple', marker='o', label='Vacuum Pump Time')
ax1.set_ylabel('Pump Time (Seconds)')
ax1.set_title('ARL 9900 Predictive Maintenance Dashboard')
ax1.axhline(y=120, color='red', linestyle='--', label='Fatal Error Threshold')
ax1.grid(True)
ax1.legend()

# Highlight the anomaly on the top plot
anomaly_data = clinker_df[clinker_df['Is_Anomaly_Multi'] == True]
ax1.scatter(anomaly_data.index, anomaly_data['SmplTime'], color='red', s=150, zorder=5, label='AI Alert')

# Bottom Plot: Chemical Drift (CaO)
ax2.plot(clinker_df.index, clinker_df['CaO'], color='blue', marker='o', label='Clinker CaO')
ax2.set_ylabel('CaO Percentage (%)')
ax2.set_xlabel('Date')
ax2.grid(True)
ax2.legend()

# Highlight the anomaly on the bottom plot
ax2.scatter(anomaly_data.index, anomaly_data['CaO'], color='red', s=150, zorder=5)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("--- FINAL AI ALERT REPORT ---")
print(anomaly_data[['CaO', 'SmplTime', 'WaterTemp']].to_string())
```
